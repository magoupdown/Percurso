```python
#@title ▶ Inicializar Percurso (executar uma vez por sessão)
import subprocess, sys, os
REPO = "https://github.com/magoupdown/Percurso.git"
try:
    if not os.path.exists("/content/Percurso_Revisao"):
        subprocess.run(["git", "clone", "--depth", "1", "--branch", "melhoria/experiencia-e-integridade", REPO, "/content/Percurso_Revisao"], check=True)
    else:
        subprocess.run(["git", "-C", "/content/Percurso_Revisao", "pull", "--ff-only", "-q"], check=True)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-c", "/content/Percurso_Revisao/requirements.txt", "-e", "/content/Percurso_Revisao"], check=True)
    # a instalação editável só é vista pelo interpretador após reiniciar; por isso o caminho entra aqui também
    if "/content/Percurso_Revisao" not in sys.path:
        sys.path.insert(0, "/content/Percurso_Revisao")
    import importlib; importlib.invalidate_caches()
    if "percurso" in sys.modules:
        print("✓ Arquivos atualizados. Se já abriu o aplicativo nesta sessão, reinicie a sessão do Colab para carregar a atualização.")
    else:
        print("✓ Percurso pronto. Execute a célula abaixo.")
except subprocess.CalledProcessError:
    print("Não foi possível atualizar ou instalar o Percurso. Confira a conexão e as mensagens acima; depois tente novamente. Seus dados no Drive foram preservados.")
```

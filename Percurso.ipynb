**Como usar:** execute a célula 1 (uma vez por sessão) e depois a célula 2. O Percurso pedirá permissão para usar uma pasta própria no seu Google Drive e abrirá a interface aqui embaixo.

Você não precisa editar nada. Se algo der errado, execute a célula novamente.

```python
#@title ▶ Inicializar Percurso (executar uma vez por sessão)
import subprocess, sys, os
REPO = "https://github.com/magoupdown/Percurso.git"
try:
    if not os.path.exists("/content/Percurso"):
        subprocess.run(["git", "clone", "--depth", "1", REPO, "/content/Percurso"], check=True)
    else:
        subprocess.run(["git", "-C", "/content/Percurso", "pull", "-q"], check=False)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", "/content/Percurso"], check=True)
    # a instalação editável só é vista pelo interpretador após reiniciar; por isso o caminho entra aqui também
    if "/content/Percurso" not in sys.path:
        sys.path.insert(0, "/content/Percurso")
    import importlib; importlib.invalidate_caches()
    print("✓ Percurso pronto. Execute a célula abaixo.")
except subprocess.CalledProcessError:
    print("Não foi possível baixar o Percurso agora (problema de rede). Aguarde alguns segundos e execute esta célula novamente.")
```
